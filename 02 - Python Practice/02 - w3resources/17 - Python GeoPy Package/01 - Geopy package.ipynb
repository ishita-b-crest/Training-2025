#### **earch the Street address, name from a given location**

```python
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut, GeocoderUnavailable
import time
```

```python
geolocator = Nominatim(user_agent="ishita.geopy.app/1.0 (contact: ishita@example.com)", timeout=10)
```

```python
ladd1 = "27488 Stanford Avenue, North Dakota"
```

```python
print("Location Address: ", ladd1)
```

```python

def get_location(address):
    try:
        location = geolocator.geocode(address)
        if location:
            print("Street address, street name:", location.address)
            print("Coordinates:", (location.latitude, location.longitude))
        else:
            print("Location not found.")
    except GeocoderTimedOut:
        print("Timeout occurred while trying to fetch:", address)
    except GeocoderUnavailable:
        print("Geocoder service unavailable. Please check your internet connection or try again later.")
    except Exception as e:
        print("Error:", e)

# Address 1
ladd1 = "27488 Stanford Avenue, North Dakota"
print("Location address:", ladd1)
get_location(ladd1)
time.sleep(1)

# Address 2
ladd2 = "380 New York St, Redlands, CA 92373"
print("\nLocation address:", ladd2)
get_location(ladd2)
time.sleep(1)

# Address 3
ladd3 = "1600 Pennsylvania Avenue NW"
print("\nLocation address:", ladd3)
get_location(ladd3)
```
